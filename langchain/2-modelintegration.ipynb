### Google Gemini 2.5 Flash Model Integration with LangChain

## Streaming And Batch

## Streaming

Most models can stream there outrput content while it is been generated. By displaying output progressively, streaming significantly improves user experience, particularly for longer responses. Calling stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real-time.

```python
model.invoke("Write me a 200 words paragraph on Artificial Intelligence")
```

```python
for chunk in model.stream("Write me a 200 words paragraph on Artificial Intelligence"):
    print(chunk.text, end="|", flush=True)
```

## Batch

Batching a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel.

```python
responses = model.batch([
    "What is the capital of France?",
    "What is the capital of India?",
    "What is the capital of Germany?"
])
for response in responses:
    print(response)
```

```python
model.batch([
    "What is the capital of France?",
    "What is the capital of India?",
    "What is the capital of Germany?"
],
config={
    'max_concurrency': 5,   # Limit to 5 parallel calls
})
```
